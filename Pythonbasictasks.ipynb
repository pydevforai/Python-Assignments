## Task 1: Cinema Ticket Price

### Task Description

User ki age input lein aur age ke according cinema ticket price display karein.

### Rules

- Age below 5 or above 100: Free
- Age 5–12: Rs. 300
- Age 13–59: Rs. 700
- Age 60-100: Rs. 400

### Requirements

1. User se name aur age input lein.
2. Age ke according ticket price calculate karein.
3. Customer name aur final ticket price display karein.

````python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

if age < 5 or age > 100:
  print("Free")

elif age >= 5 and age <= 12:
  price = "Rs. 300"

elif age >= 13 and age <= 59:
  price = "Rs. 700"

elif age >= 60 and age <= 100:
  price = "Rs. 400"

print(f"Customer Name: {name}")
print(f"Ticket Price: {price}")
````

## Task 2: Temperature Converter

### Task Description

User se Celsius temperature input lein aur usko Fahrenheit mein convert karein. Temperature ka weather message bhi display karein.

### Formula

```
Fahrenheit = (Celsius × 9/5) + 32
```

### Weather Rules

- Below 10°C: Very Cold
- 10°C–19°C: Cold
- 20°C–29°C: Normal
- 30°C or above: Hot

### Requirements

1. Celsius temperature input lein.
2. Fahrenheit calculate karein.
3. Weather condition display karein.

````python
Temperature = int(input("Enter Temperature in Celsius: "))
Fahrenheit = (Temperature * 9/5) + 32

if Temperature < 10:
  weather = "Very Cold"

elif Temperature >= 10 and Temperature <= 19:
  weather = "Cold"

elif Temperature >= 20 and Temperature <= 29:
  weather = "Normal"

else:
  weather = "Hot"

print(f"Temperature in Fahrenheit: {Fahrenheit}°F")
print(f"Weather: {weather}")
````

## Task 4: Weekly Steps Tracker

### Task Description

User se seven days ke walking steps input lein. Total aur average steps calculate karein.

### Requirements

1. `for` loop seven times run karein.
2. Har din ke steps input lein.
3. Total steps calculate karein.
4. Daily average calculate karein.
5. Agar average `5,000` ya above ho to `Active Week` display karein, warna `Needs More Activity`.

````python
total_steps = 0
for i in range(7):
  steps = int(input("Enter your steps: "))
  total_steps += steps
  daily_avg = total_steps/7

print(f'Total Steps: {total_steps}')
print(f'Daily Average: {daily_avg}')

if daily_avg >= 5000:
    print("Active Week")

else:
    print("Need More Activity")
````

## Task 5: Secure PIN Checker

### Task Description

Ek secret PIN program mein save karein. User ko correct PIN enter karne ke liye maximum three attempts dein.

### Requirements

1. Secret PIN `1234` save karein.
2. User se PIN input lein.
3. Correct PIN par `Access Granted` display karein.
4. Incorrect PIN par remaining attempts show karein.
5. Three incorrect attempts ke baad `Account Blocked` display karein.

````python
secret_pin = 2345
user_input = int(input("Enter your pin: "))

for i in range(1, 4):
  if user_input == secret_pin:
    print("Access Granted")
    break

  else:
    print("Access Denied")
    print(f"No of Attempts: {i}")
    user_input = int(input("Enter your pin: "))
    if i == 3:
      print("Account Blocked")
      break
````

````python
menu= {"Burger":500,
       "Pizza":1200,
       "Pasta":800,
       "Drink":200
}

orders = []

print("========= Restaurant Menu =========")

for key, value in menu.items():
  print(f"{key} : {value}")

while True:
   item_name = input("Enter item name: ").title()
   if item_name in menu:
    quantity = int(input("Enter quantity: "))
    price = menu[item_name]
    sub_total = price * quantity

    orders.append({
        "item": item_name,
        "quantity": quantity,
        "sub_total": sub_total
    })

   else:
    print("Item is not available")

   choice = input("You want to add more items (yes/no)")
   if choice == "no":
    break

total = 0
for order in orders:
  total += order["sub_total"]
  print("Total:",total)

tax = total * 0.13

if total >= 3000:
  discount = total * 0.1

else:
  discount = 0

final_total = total + tax - discount

print("========= Receipt =========")
print(f"Total: {total}")
print(f"Tax: {tax}")
print(f"Discount: {discount}")
print(f"Final Total: {final_total}")

print(orders)
````
